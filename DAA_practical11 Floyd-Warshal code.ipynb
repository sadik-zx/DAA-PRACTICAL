```python

n = int(input("Enter number of vertices: "))

dist = []

print("Enter adjacency matrix (-1 for no edge):")

for i in range(n):
    row = list(map(int, input().split()))

    for j in range(n):
        if row[j] == -1:
            row[j] = 99999

    dist.append(row)

# Floyd-Warshall Algorithm
for k in range(n):
    for i in range(n):
        for j in range(n):

            if dist[i][k] != 99999 and dist[k][j] != 99999:
                if dist[i][k] + dist[k][j] < dist[i][j]:
                    dist[i][j] = dist[i][k] + dist[k][j]

# Display shortest distance matrix
print("Shortest distance matrix:")

for i in range(n):
    for j in range(n):
        if dist[i][j] == 99999:
            print("INF", end=" ")
        else:
            print(dist[i][j], end=" ")
    print()
```
