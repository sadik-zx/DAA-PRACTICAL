```python
def prim(graph, n):
    visited = [False] * n
    visited[0] = True

    edges = []
    totalcost = 0

    for i in range(n - 1):
        mincost = float('inf')
        u = -1
        v = -1

        for i in range(n):
            if visited[i]:
                for j in range(n):
                    if not visited[j] and graph[i][j] != 0:
                        if graph[i][j] < mincost:
                            mincost = graph[i][j]
                            u = i
                            v = j

        if u == -1:
            print("Graph is not connected!")
            return

        edges.append((u, v, mincost))
        totalcost += mincost
        visited[v] = True

    print("Minimum Spanning Tree:")
    for u, v, cost in edges:
        print(u, "-", v, ":", cost)

    print("Total Minimum Cost:", totalcost)


n = int(input("Enter number of vertices: "))
graph = [[0] * n for i in range(n)]

e = int(input("Enter number of edges: "))

print("Enter edges as: source destination weight")
for i in range(e):
    u, v, w = map(int, input().split())
    graph[u][v] = w
    graph[v][u] = w

prim(graph, n)
```
