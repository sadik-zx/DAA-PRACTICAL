```python
def knapsack(W, wt, val, n):
    dp = [[0] * (W + 1)
    for i in range(n + 1)]

    for i in range(1, n + 1):
        for w in range(1, W + 1):

            if wt[i - 1] <= w:
                dp[i][w] = max(
                    val[i - 1] + dp[i - 1][w - wt[i - 1]],
                    dp[i - 1][w]
                )
            else:
                dp[i][w] = dp[i - 1][w]

    return dp[n][W]



W = 8
wt = [2, 3, 4, 5]
val = [1, 2, 5, 6]

n = len(wt)

print("Maximum Profit =", knapsack(W, wt, val, n))
```
