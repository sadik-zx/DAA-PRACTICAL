```python

from itertools import permutations

n = int(input("Enter number of cities: "))

cost = []

print("Enter cost matrix:")

for i in range(n):
    row = list(map(int, input().split()))
    cost.append(row)

# Generate all possible paths
path = list(range(1, n))

min cost = float('inf')
best path = []

for p in permutations(path):

    current cost = cost[0][p[0]]

    for i in range(n - 2):
        current cost += cost[p[i]][p[i + 1]]

    current cost += cost[p[n - 2]][0]

    if current cost < min cost:
        min cost = current cost
        best path = [0] + list(p)

print("Minimum cost:", min cost)

print("Optimal tour:", end=" ")

for city in best path:
    print(city, "->", end=" ")

print("0")
```
