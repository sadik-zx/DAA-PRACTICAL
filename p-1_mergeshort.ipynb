```python
def merge_sort(a):
    if len(a) <= 1:
        return a
    mid = len(a)//2
    l = merge_sort(a[:mid])
    r = merge_sort(a[mid:])
    res = []
    while l and r:
        res.append(l.pop(0) if l[0] < r[0] else r.pop(0))
    return res + l + r

n = int(input("Enter size: "))
a = list(map(int, input("Enter elements: ").split()))
print("Sorted:", merge_sort(a))
```
