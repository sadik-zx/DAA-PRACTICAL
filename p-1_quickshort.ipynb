```python
def quicksort(a):
    if len(a) <= 1:
        return a
    p = a[0]
    left = [x for x in a[1:] if x <= p]
    right = [x for x in a[1:] if x > p]
    return quicksort(left) + [p] + quicksort(right)

n = int(input("Enter size: "))
a = list(map(int, input("Enter elements: ").split()))

print("Sorted:", quicksort(a))
```
