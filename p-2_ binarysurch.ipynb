```python
a = list(map(int, input("Enter elements: ").split()))
a.sort()
key = int(input("Enter key: "))

low = 0
high = len(a)-1

while low <= high:
    mid = (low+high)//2

    if a[mid] == key:
        print("Found at position:", mid+1)
        break
    elif a[mid] < key:
        low = mid+1
    else:
        high = mid-1
else:
    print("Not Found")
```
